```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Statistical Summary:")
display(df.describe())
```

```python
categorical_columns = [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]

for column in categorical_columns:
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    print(df[column].value_counts())
```

```python
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

display(missing_table)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(df.shape)
```

```python

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())


categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

```python
print(df.dtypes)
```

```python
print("Checking numerical columns for negative values:\n")

for column in numeric_columns:
    negative_count = (df[column] < 0).sum()

    if negative_count > 0:
        print(column, ":", negative_count)
```

```python
Q1 = df["Yield_Tonnes_Ha"].quantile(0.25)
Q3 = df["Yield_Tonnes_Ha"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

yield_outliers = df[
    (df["Yield_Tonnes_Ha"] < lower_limit) |
    (df["Yield_Tonnes_Ha"] > upper_limit)
]

print("Number of yield outliers:", len(yield_outliers))

display(
    yield_outliers[
        [
            "Farm_ID",
            "State",
            "District",
            "Crop",
            "Season",
            "Yield_Tonnes_Ha"
        ]
    ].head(20)
)
```

```python
season_results = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Profit_INR": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
})

best_yield_season = season_results[
    "Yield_Tonnes_Ha"
].idxmax()

worst_yield_season = season_results[
    "Yield_Tonnes_Ha"
].idxmin()

best_profit_season = season_results[
    "Profit_INR"
].idxmax()

best_water_season = season_results[
    "Water_Efficiency_t_per_1000m3"
].idxmax()

highest_risk_season = season_results[
    "Disease_Pest_Risk_pct"
].idxmax()

print("KEY OBSERVATIONS")
print("=" * 65)

print(
    f"Highest average yield: {best_yield_season} "
    f"({season_results.loc[best_yield_season, 'Yield_Tonnes_Ha']:.2f} Tonnes/Ha)"
)

print(
    f"Lowest average yield: {worst_yield_season} "
    f"({season_results.loc[worst_yield_season, 'Yield_Tonnes_Ha']:.2f} Tonnes/Ha)"
)

print(
    f"Highest average profit: {best_profit_season} "
    f"(₹{season_results.loc[best_profit_season, 'Profit_INR']:,.2f})"
)

print(
    f"Highest water efficiency: {best_water_season} "
    f"({season_results.loc[best_water_season, 'Water_Efficiency_t_per_1000m3']:.2f})"
)

print(
    f"Highest disease/pest risk: {highest_risk_season} "
    f"({season_results.loc[highest_risk_season, 'Disease_Pest_Risk_pct']:.2f}%)"
)
```

```python
final_table = df.groupby("Season").agg(
    Average_Yield_Tonnes_Ha=(
        "Yield_Tonnes_Ha",
        "mean"
    ),

    Average_Production_Tonnes=(
        "Production_Tonnes",
        "mean"
    ),

    Average_Rainfall_mm=(
        "Rainfall_mm",
        "mean"
    ),

    Average_Temperature_C=(
        "Avg_Temperature_C",
        "mean"
    ),

    Average_Water_Used_m3=(
        "Water_Used_m3",
        "mean"
    ),

    Average_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    ),

    Average_Revenue_INR=(
        "Revenue_INR",
        "mean"
    ),

    Average_Cost_INR=(
        "Total_Cost_INR",
        "mean"
    ),

    Average_Profit_INR=(
        "Profit_INR",
        "mean"
    ),

    Average_Disease_Pest_Risk=(
        "Disease_Pest_Risk_pct",
        "mean"
    )
)

display(final_table.round(2))
```
